## 1. Chọn sáu thuộc tính và tổng kết
Đầu vào là bản 299 hồ sơ sau xử lý nhãn từ notebook 17. Không dùng tập chỉ 298 nhãn hợp lệ vì bước này không cần nhãn. MaHoSo là khóa liên kết, không tính vào số chiều.

```python
from pathlib import Path
from decimal import Decimal
import pandas as pd

thu_muc = Path.cwd()
if thu_muc.name == "python":
    thu_muc = thu_muc.parent
file_vao = thu_muc / "ket_qua/15_khong_nhat_quan/02_death_event/ban_lam_viec_nhan_python.csv"
if not file_vao.exists():
    raise FileNotFoundError("Chạy Run All notebook 17_xu_ly_death_event.ipynb trước.")
ban_day_du = pd.read_csv(file_vao, dtype=str, keep_default_na=False)
cac_cot_giu = ["age", "creatinine_phosphokinase", "ejection_fraction",
               "platelets", "serum_creatinine", "serum_sodium"]
bang_giam_chieu = ban_day_du[["MaHoSo"] + cac_cot_giu].copy()
so_cot_goc = len([c for c in ban_day_du.columns if c not in ["MaHoSo", "age_group"]])
so_dau_vao = so_cot_goc - 1  # Không tính nhãn DEATH_EVENT
tong_ket = pd.DataFrame({"SoHoSoTruoc":[len(ban_day_du)],
    "SoHoSoSau":[len(bang_giam_chieu)], "SoCotGocGomNhan":[so_cot_goc],
    "SoThuocTinhDauVao":[so_dau_vao], "SoThuocTinhGiuLai":[len(cac_cot_giu)],
    "TyLeGiamPhanTram":[Decimal(100) * (so_dau_vao - len(cac_cot_giu)) / so_dau_vao]})
tong_ket
```

## 2. Danh sách sáu thuộc tính
Lược khỏi bảng con 5 mã danh mục và time. Nhãn vẫn ở bảng đầy đủ. Bảng con có 7 cột vật lý nếu tính cả MaHoSo.

```python
thuoc_tinh = pd.DataFrame({"STT":range(1, len(cac_cot_giu)+1),
                           "ThuocTinhGiuLai":cac_cot_giu})
thuoc_tinh
```

## 3. Năm hồ sơ minh họa
CPK và EF là tên rút gọn để hiển thị. Chỉ bảng minh họa được định dạng; bảng lưu giữ nguyên chuỗi giá trị. MaHoSo nội bộ có thể khác SQL nên năm ví dụ có thể thuộc các hồ sơ khác.

```python
vi_du = bang_giam_chieu.copy()
vi_du["MaHoSo"] = vi_du["MaHoSo"].astype(int)
vi_du = vi_du.sort_values("MaHoSo").head(5).reset_index(drop=True)
for cot in cac_cot_giu:
    vi_du[cot] = vi_du[cot].map(Decimal)
vi_du = vi_du.rename(columns={"creatinine_phosphokinase":"CPK", "ejection_fraction":"EF"})
vi_du
```

**Nhận xét:** Giữ 299 hồ sơ; giảm 12 thuộc tính đầu vào xuống 6, tương ứng 50%. Giá trị của sáu biến giữ nguyên. Bảng con phục vụ thống kê định lượng, không thay thế bảng đầy đủ khi cần giới tính, tiền sử, hút thuốc, thời gian theo dõi hay nhãn kết quả. Đây mới là trường hợp 1 trong tối thiểu 2 trường hợp của đề.

## 4. Lưu kết quả riêng
Không ghi đè nguồn hoặc thay bảng chính bằng bảng giảm chiều.

```python
thu_muc_kq = thu_muc / "ket_qua/16_giam_thieu/01_giam_so_chieu"
thu_muc_kq.mkdir(parents=True, exist_ok=True)
tong_ket.to_csv(thu_muc_kq / "tong_ket_python.csv", index=False, encoding="utf-8-sig")
thuoc_tinh.to_csv(thu_muc_kq / "thuoc_tinh_python.csv", index=False, encoding="utf-8-sig")
bang_giam_chieu.to_csv(thu_muc_kq / "bang_giam_chieu_python.csv", index=False, encoding="utf-8-sig")
pd.DataFrame({"TrangThai":["Đã lưu bảng con sáu thuộc tính; giữ nguyên bảng đầy đủ."]})
```
