Đọc CSV before bằng Python và kiểm tra hai chỉ tiêu tương ứng với SQL: tổng số dòng và số ô chứa `Invalid_Age`. Bước này chưa làm sạch dữ liệu.

```python
from pathlib import Path
import pandas as pd

# Xác định thư mục đồ án và đọc nguyên trạng CSV before
thu_muc = Path.cwd()
if thu_muc.name == "python":
    thu_muc = thu_muc.parent
file_csv = thu_muc / "VoMinhHuy_Heart Failure Prediction_before.csv"
df = pd.read_csv(file_csv, dtype=str, keep_default_na=False)

# Đối chiếu hai chỉ tiêu đã kiểm tra bằng SQL
ket_qua = pd.DataFrame({
    "TongSoDong": [len(df)],
    "SoOInvalidAge": [df["age"].eq("Invalid_Age").sum()]
})
ket_qua
```

**Nhận xét:** Python ghi nhận **309 dòng** và **1 ô `Invalid_Age`**, trùng với hai kết quả đã chạy trong SQL Server. Đây là kiểm tra số lượng và bảo toàn giá trị sai kiểu, chưa phải so sánh toàn bộ từng ô.
